```python
#defining a temperature conversion to Fahrenheit degree function
def celsiusconversion(tempC):
    tempF=tempC*1.8+32
    return tempF

#asking number of repetition for the loop
n = int(input("How many temperatures do you want to convert? "))

#the loop
for i in range(n):

  # user enter the temperature in celsius degree
  tempC=input("Enter the temperature in °C:")

  #Conversion of the text entered by the user, to a decimal number
  tempC=float(tempC)

  #Conversion to the temperature using the function celsiusconversion
  tempF=celsiusconversion(tempC)

  #Showing the conversion
  print(f"{tempC} °C =  {tempF} °F")

  #reporting whether the temperature is above or below the threshold of boiling point water
  if tempC > 100:
      print("The temperature is above the boiling point of water.")
  elif tempC < 100:
      print("The temperature is below the boiling point of water.")
  else:
      print("The temperature is at the boiling point of water.")

```
