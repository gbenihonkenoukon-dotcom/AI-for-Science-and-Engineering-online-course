```python
#2-storing of dataset file in dataframe df
import pandas as pd
df = pd.read_excel("Dataset for training_Organized.xlsx")

df.head()
```

```python
#3-calculations

df = df.select_dtypes(include="number") # considering only numerical values (without column header) as df

stat = pd.DataFrame({
    "Mean": df.mean(),
    "Median": df.median(),
    "Range": df.max() - df.min(),
    "Variance": df.var(),
    "Standard deviation": df.std()
})

# Displaying the results
stat
```

Covariance and correlation between stress and strain

```python
# Calculation
covariance = df["stress"].cov(df["strain"])

# Calculate correlation
correlation = df["stress"].corr(df["strain"])

print("Covariance:", covariance)
print("Correlation:", correlation)
```

Covariance and correlation between void ratio and strain

```python
# Calculation
covariance = df["void ratio"].cov(df["strain"])

# Calculate correlation
correlation = df["void ratio"].corr(df["strain"])

print("Covariance:", covariance)
print("Correlation:", correlation)
```

**Explanation of statistics**

- Stress: The mean is 148.26 and the standard deviation is 91.66, indicating considerable variation. The mean exceeds the median (120.10), which may suggest that high stress values influence the distribution.

- Strain: The mean is 0.1929 and the standard deviation is 0.1050. The mean and median (0.1920) are very close, suggesting similar central values.

- Void ratio: The mean is 2.0442 and the standard deviation is 0.4046. The mean and median (2.0310) are close, while the range of 1.6355 indicates variation in void ratio across the dataset.

**Explanation of correlation**

1. Stress and strain

Correlation = 0.8715: This indicates a strong positive linear relationship. Higher stress values are generally associated with higher strain values (soil deformation).

This is consistent with the general behaviour of soil under loading.

2. Void ratio and strain

Correlation = −0.9342: This indicates a very strong negative linear relationship. Higher strain values are generally associated with lower void ratios.

This relationship is consistent with soil compressive deformation. As clay compresses, its void ratio decreases.

***No, correlation does not necessarily mean that one variable causes the other***

It just describes an observed association. Correlation tells both the direction and the strength of a linear relationship between the variables. And it can also be affected by outliers.

**Choosing stress as input and strain as output, a simple mathematical expression could be:**

Stress (σ) ⟶ f(σ) ⟶ Strain (ε)

Mathematically, the model can be expressed as:

ε=f(σ)

Where:

* Input (σ): Stress.

* Function/model (f): The machine-learning model that learns the relationship between stress and strain from the dataset.

* Output (ε): Predicted strain.
